## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

We are doing unsupervised clustering (Lane 3). As we have tens of thousands of web pages, we are dividing them into distinct behavioral clusters. This helps the team apply the same strategic playbook/action to all web pages belonging to that specific cluster.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

We do not have a target column in the data. The model predicts a discrete cluster for each web page. This label is derived mathematically from the multivariate feature space of observed measurements (impressions, CTR, ranking position, and engagement) rather than an arbitrary human-defined product rule.

## 3. Success metric

*One metric you can defend. What number means 'good'?*

1.Mathematical Metric: Silhouette Score. We aim for a score > 0.40, proving that clusters are mathematically dense and well-separated from one another.

2.Business Actionability Metric: Cluster Cohesion. We evaluate whether the vast majority (≥80%) of pages in each cluster share the same underlying bottleneck so the same strategic remedy applies across the group.

## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

**Unit of Analysis (Grain):** One row = One unique published content item / web page (`content_id`).

```python
import pandas as pd
try:
    df = pd.read_csv("/content/content_refresh_anonymized.csv")
except FileNotFoundError:
    df = pd.read_csv("data/raw/content_refresh_anonymized.csv")
display_cols = [
    "content_id",
    "client_id",
    "impressions_90d",
    "avg_position",
    "ctr",
    "engagement_rate",
    "content_age_days",
    "word_count",
]
print(f"Dataset Shape: {df.shape[0]:,} rows x {df.shape[1]} columns")
print(
    "Unit of Analysis: 1 Row = 1 Unique Web Page"
    f" ({df['content_id'].nunique():,} unique pages)\n"
)
# Display the top 5 rows
df[display_cols].head(5)
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule fails because:

1.Multi-Dimensional Interaction: A page's performance is governed by continuous interactions across 9 metrics (impressions, ranking position, CTR, engagement, word count, age). Writing rules for all interacting combinations creates hundreds of brittle IF-statements.

2.Rigid Cutoffs vs. Continuous Space: A rule with impressions >= 500 treats 499 impressions as completely different from 501. Clustering calculates continuous Euclidean distance smoothly.

3.Multiple Websites: Across 32 different client websites, traffic levels differ. Clustering normalizes the data to adapt to each site's distribution automatically.

```python
summary_stats = df[[
    "impressions_90d",
    "avg_position",
    "ctr",
    "engagement_rate",
    "word_count",
]].describe()
print(summary_stats.round(2))
```
